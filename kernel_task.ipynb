```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import convolve
```

```python
input_path = "image.png"
img = cv2.imread(input_path,cv2.IMREAD_GRAYSCALE)
plt.figure(figsize=(6,4))
plt.imshow(img,cmap='gray')
#plt.show()
```

```python
sharpen = [[0, -1, 0],
           [-1, 5, -1],
           [0, -1, 0]]

kernel = sharpen
im = np.array(img, dtype=float)
img2 = convolve(im, kernel)

plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2, cmap='gray')
plt.show()
```

```python
blur = np.ones((3,3), dtype=float)/9

kernel = blur
im = np.array(img, dtype=float)
img2 = convolve(im, kernel)

plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2, cmap='gray')
plt.show()
```

```python
custom = [[0, 1, 0],
           [1, -1, 1],
           [0, 1, 0]]

kernel =custom
im = np.array(img, dtype=float)
img2 = convolve(im, kernel)

plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2, cmap='gray')
plt.show()
```

```python
random = [[1, 0, -1],
          [2, 1, 0],
          [0, -1, 2]]

kernel = random
im = np.array(img, dtype=float)
img2 = convolve(im, kernel)

plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2, cmap='gray')
plt.show()
```

```python
blur = [[1, 2, 1],
        [2, 4, 2],
        [1, 2, 1]]

kernel = np.array(blur, dtype=float)/16
im = np.array(img, dtype=float)
img2 = convolve(im, kernel)

plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2, cmap='gray')
plt.show()
```
